La variable `TotalVenta` representa el valor de cada registro comercial y se calculó multiplicando `Quantity` por `UnitPrice`.

En esta comprobación se verificará si los valores almacenados coinciden con el resultado de dicha operación. Esto permitirá detectar posibles inconsistencias antes de utilizar esta variable para calcular los indicadores de ventas.

### **8.3. Análisis de las variables comerciales**

Para profundizar en el comportamiento de las transacciones, se calcularán medidas estadísticas específicas para `Quantity`, `UnitPrice` y `TotalVenta`.

Estas variables permiten estudiar cuántas unidades aparecen registradas en las transacciones, cómo se distribuyen los precios unitarios y qué valores presentan las ventas individuales. ### **8.4. Indicadores generales**

En esta sección se calcularán algunos indicadores que permiten resumir el comportamiento comercial del conjunto de datos, como el número de transacciones registradas, las unidades acumuladas, el valor total de las ventas y el promedio de venta por registro.

Estos indicadores ofrecen una primera aproximación al volumen de actividad comercial registrado en los datos procesados.

Para evitar confusiones, el valor total calculado corresponde a la suma de `TotalVenta` en los registros conservados después del ETL. Por lo tanto, describe el conjunto de datos procesado y no necesariamente representa la totalidad de los movimientos originales de la empresa.
***

```python
# Calcular indicadores generales del conjunto de datos

cantidad_registros = len(df)
unidades_totales = df["Quantity"].sum()
ventas_totales = df["TotalVenta"].sum()
venta_promedio = df["TotalVenta"].mean()
venta_mediana = df["TotalVenta"].median()

print("INDICADORES GENERALES")
print("-" * 40)
print(f"Registros analizados: {cantidad_registros:,}")
print(f"Unidades registradas: {unidades_totales:,}")
print(f"Valor total de las ventas: {ventas_totales:,.2f}")
print(f"Venta promedio por registro: {venta_promedio:,.2f}")
print(f"Mediana del valor de venta: {venta_mediana:,.2f}")
```

Los resultados estadísticos permiten describir el comportamiento general de las transacciones comerciales. La comparación entre la media y la mediana ayuda a reconocer si los valores presentan diferencias importantes, mientras que la desviación estándar permite evaluar la variabilidad de las cantidades y los precios.

## **9. Análisis descriptivo de las ventas**

En esta sección se estudiará el comportamiento comercial del conjunto de datos mediante comparaciones por periodos de tiempo, productos y países. El objetivo es identificar patrones que ayuden a comprender cómo se distribuyen las ventas y cuáles son los elementos que tienen mayor participación en la actividad comercial.

Para realizar estos análisis se utilizará la variable `TotalVenta`, creada durante la actividad anterior de ETL, junto con las variables de fecha, cantidad, descripción del producto y país.
***

### **9.1. Preparación de las fechas**

Para estudiar el comportamiento de las ventas a través del tiempo, se verificará que la variable `InvoiceDate` tenga un formato de fecha adecuado. A partir de esta variable se podrá agrupar la información por mes y año, lo que facilitará la identificación de periodos con mayores o menores ventas.
***

```python
# Convertir InvoiceDate al formato de fecha
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

# Comprobar el tipo de dato y las fechas que no pudieron convertirse
print("Tipo de dato de InvoiceDate:", df["InvoiceDate"].dtype)
print("Fechas no válidas:", df["InvoiceDate"].isna().sum())
```

### **9.2. Análisis de las ventas mensuales**

El análisis mensual permite comparar el valor de las ventas registradas en diferentes periodos y observar cómo cambia la actividad comercial a lo largo del tiempo.

Para ello, se agruparán los registros por año y mes y se sumará la variable `TotalVenta`. Posteriormente, se representarán los resultados mediante una gráfica de líneas, que facilitará la identificación de aumentos, disminuciones y posibles patrones temporales.

Este análisis describe los registros disponibles en el dataset procesado y no permite afirmar por sí solo cuáles son las causas de los cambios observados.
***

```python
# Crear una copia con los registros que tienen fecha válida
df_fechas = df.dropna(subset=["InvoiceDate"]).copy()

# Crear una columna que identifique cada mes y año
df_fechas["Periodo"] = df_fechas["InvoiceDate"].dt.to_period("M")

# Agrupar los registros por mes y sumar el valor de las ventas
ventas_mensuales = (
    df_fechas.groupby("Periodo")["TotalVenta"]
    .sum()
    .reset_index()
)

# Convertir el periodo a texto para facilitar la visualización
ventas_mensuales["Periodo"] = ventas_mensuales["Periodo"].astype(str)

# Mostrar la tabla de resultados
display(ventas_mensuales)
```

```python
# Graficar el comportamiento de las ventas por mes

plt.figure(figsize=(12, 5))

sns.lineplot(
    data=ventas_mensuales,
    x="Periodo",
    y="TotalVenta",
    marker="o"
)

plt.title("Comportamiento de las ventas por mes")
plt.xlabel("Mes")
plt.ylabel("Valor total de las ventas")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### **9.3. Análisis de los productos más vendidos**

Conocer los productos con mayor cantidad de unidades vendidas puede ayudar a una empresa a reconocer cuáles tienen mayor movimiento comercial. Esta información puede ser útil para apoyar decisiones relacionadas con el inventario, la disponibilidad de productos y la planificación de las compras.

En este análisis se agruparán los registros por descripción del producto y se sumarán las cantidades. Luego se seleccionarán los diez productos con mayores cantidades acumuladas para representarlos en una gráfica de barras horizontales.

El resultado representa las unidades registradas en el archivo procesado, no el número de facturas en las que aparece cada producto.
***

```python
# Agrupar por producto y sumar las unidades registradas

unidades_por_producto = (
    df.groupby("Description")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

# Seleccionar los diez productos con mayor cantidad acumulada
top_productos = unidades_por_producto.head(10).sort_values()

# Mostrar la tabla
display(top_productos.to_frame(name="Unidades vendidas"))
```

```python
# Visualizar los diez productos con mayor cantidad de unidades

plt.figure(figsize=(11, 6))

top_productos.plot(kind="barh")

plt.title("Top 10 de productos por unidades vendidas")
plt.xlabel("Unidades registradas")
plt.ylabel("Producto")
plt.tight_layout()
plt.show()
```

### **9.4. Análisis de las ventas por país**

La comparación de las ventas por país permite observar cómo se distribuye el valor comercial entre los distintos mercados incluidos en el conjunto de datos.

Para realizar este análisis se agruparán los registros por país y se sumará la variable `TotalVenta`. Después, se seleccionarán los diez países con mayores valores acumulados y se representarán mediante una gráfica de barras horizontales.

Este resultado permite identificar los países con mayor valor de ventas registrado en los datos procesados. Sin embargo, no demuestra por sí solo que esos mercados sean los más rentables, ya que para calcular rentabilidad también sería necesario conocer los costos y otros gastos asociados.
***

```python
# Agrupar los registros por país y sumar el valor de las ventas

ventas_por_pais = (
    df.groupby("Country")["TotalVenta"]
    .sum()
    .sort_values(ascending=False)
)

# Seleccionar los diez países con mayores ventas
top_paises = ventas_por_pais.head(10).sort_values()

# Mostrar la tabla de resultados
display(top_paises.to_frame(name="Valor total de ventas").round(2))
```

# **11. Comparación entre unidades vendidas e ingresos por producto**

## 11.1. Objetivo de la comparación

En esta sección se compararán las unidades registradas y el valor acumulado de las ventas por producto. El objetivo es identificar si los productos con mayor cantidad de unidades también se encuentran entre los que generan mayores ingresos.

Para ello, se agruparán los datos por descripción del producto y se calcularán dos indicadores: la suma de `Quantity` y la suma de `TotalVenta`. Posteriormente, se identificarán los diez productos con mayores ingresos y se compararán sus resultados con los productos que registraron más unidades.

Esta comparación permitirá observar dos perspectivas diferentes del comportamiento comercial: el movimiento de los productos en términos de unidades y su contribución al valor acumulado de las ventas.
***

```python
# Agrupar los datos por producto y calcular unidades e ingresos

resumen_productos = (
    df.groupby("Description")
    .agg(
        Unidades_vendidas=("Quantity", "sum"),
        Ingresos_totales=("TotalVenta", "sum")
    )
    .sort_values("Ingresos_totales", ascending=False)
)

# Mostrar los diez productos con mayores ingresos
top_ingresos = resumen_productos.head(10)

display(top_ingresos.round(2))
```

```python
# Seleccionar los diez productos con mayores ingresos
# y ordenar sus valores para facilitar la lectura de la gráfica

grafica_ingresos = (
    top_ingresos["Ingresos_totales"]
    .sort_values(ascending=True)
)

plt.figure(figsize=(11, 6))

grafica_ingresos.plot(kind="barh")

plt.title("Top 10 de productos por ingresos totales")
plt.xlabel("Valor total de las ventas")
plt.ylabel("Producto")
plt.tight_layout()
plt.show()
```

```python
# Obtener los diez productos con más unidades registradas

top_unidades = (
    resumen_productos
    .sort_values("Unidades_vendidas", ascending=False)
    .head(10)
    .reset_index()
)

# Obtener los diez productos con mayores ingresos
top_ingresos_tabla = (
    resumen_productos
    .sort_values("Ingresos_totales", ascending=False)
    .head(10)
    .reset_index()
)

# Identificar los productos que aparecen en ambas listas
productos_en_comun = set(top_unidades["Description"]) & set(
    top_ingresos_tabla["Description"]
)

print("Productos que aparecen en ambos grupos:", len(productos_en_comun))
print("\nProductos presentes en ambas listas:")

for producto in sorted(productos_en_comun):
    print("-", producto)

# Mostrar la tabla completa de los productos con mayores ingresos
display(top_ingresos_tabla.round(2))
```

### **11.6. Interpretación de los resultados**

Los resultados muestran diferencias entre los productos que generan mayores ingresos y aquellos que registran más unidades vendidas.

El producto `DOTCOM POSTAGE` presenta el mayor valor acumulado de ventas, con 206.248,77, a pesar de registrar solamente 708 unidades. En segundo lugar se encuentra `REGENCY CAKESTAND 3 TIER`, con ingresos de 174.156,54 y 13.862 unidades registradas. Por su parte, `PAPER CRAFT, LITTLE BIRDIE` registra la mayor cantidad de unidades vendidas, con 80.995, y genera ingresos de 168.469,60.

Estos resultados muestran que una cantidad elevada de unidades no implica necesariamente mayores ingresos. El valor acumulado también depende del precio unitario y de las características de las transacciones. Además, el hecho de que `DOTCOM POSTAGE` aparezca en primer lugar debe interpretarse teniendo en cuenta que corresponde a un concepto de envío y no necesariamente a un producto físico convencional.

Al comparar los dos grupos de diez productos, se encontró que cinco aparecen en ambas listas: `JUMBO BAG RED RETROSPOT`, `MEDIUM CERAMIC TOP STORAGE JAR`, `PAPER CRAFT, LITTLE BIRDIE`, `RABBIT NIGHT LIGHT` y `WHITE HANGING HEART T-LIGHT HOLDER`.

La coincidencia indica que estos productos destacan tanto por sus unidades acumuladas como por el valor de sus ventas. Desde una perspectiva empresarial, sería conveniente analizar por separado los productos con alta demanda y los productos con mayores ingresos, ya que ambos indicadores aportan información diferente para las decisiones comerciales.

Es importante recordar que los ingresos calculados corresponden a la suma de los valores registrados en el archivo procesado. No representan necesariamente la utilidad o ganancia neta, porque no se dispone de información completa sobre costos, gastos y márgenes de beneficio.

# **12. Análisis de la participación de las ventas por país**

## 12.1. Objetivo del análisis

La participación porcentual permite conocer qué proporción del valor total de las ventas corresponde a cada país. Este indicador facilita la comparación entre mercados y permite evaluar si las ventas se encuentran distribuidas de manera equilibrada o concentradas principalmente en uno de ellos.

En esta sección se calculará el valor acumulado de las ventas por país y su participación porcentual respecto al total del conjunto de datos procesado. Los resultados ayudarán a cuantificar la importancia relativa de cada mercado dentro de los registros analizados.
***

```python
# Calcular el valor de las ventas por país

resumen_paises = (
    df.groupby("Country")["TotalVenta"]
    .sum()
    .sort_values(ascending=False)
    .to_frame(name="Ventas_totales")
)

# Calcular el porcentaje que representa cada país
ventas_globales = resumen_paises["Ventas_totales"].sum()

resumen_paises["Participacion_porcentual"] = (
    resumen_paises["Ventas_totales"] / ventas_globales * 100
)

# Mostrar los diez países con mayor participación
display(resumen_paises.head(10).round(2))
```

```python
# Seleccionar los diez países con mayor participación
participacion_top10 = (
    resumen_paises["Participacion_porcentual"]
    .head(10)
    .sort_values(ascending=True)
)

# Graficar la participación porcentual
plt.figure(figsize=(11, 6))

participacion_top10.plot(kind="barh")

plt.title("Participación de los diez principales países en las ventas")
plt.xlabel("Participación porcentual (%)")
plt.ylabel("País")
plt.tight_layout()
plt.show()
```

```python
# Consultar la participación del Reino Unido

participacion_reino_unido = resumen_paises.loc[
    "United Kingdom", "Participacion_porcentual"
]

print(
    f"Participación del Reino Unido en las ventas: "
    f"{participacion_reino_unido:.2f}%"
)
```

### **12.4. Interpretación de la participación por país**

Los resultados muestran una concentración considerable del valor de las ventas en el Reino Unido, que representa el 84,59 % del total registrado en el conjunto de datos procesado. En comparación, Países Bajos representa el 2,68 %, Irlanda el 2,66 %, Alemania el 2,15 % y Francia el 1,97 %.

La diferencia entre el Reino Unido y los demás países indica que el mercado británico tiene una participación predominante dentro de las transacciones analizadas. Los otros países presentan porcentajes considerablemente menores, aunque también forman parte de la actividad comercial registrada.

Desde una perspectiva empresarial, esta información puede servir para estudiar la dependencia de un mercado principal y evaluar posibles oportunidades de crecimiento internacional. Por ejemplo, sería posible investigar qué productos tienen mayor demanda en los países con menor participación y comparar sus patrones de compra con los del Reino Unido.

Sin embargo, estos resultados no permiten afirmar que el Reino Unido sea el mercado más rentable, ya que el análisis se basa en el valor de las ventas y no incluye información suficiente sobre costos, gastos y márgenes de beneficio.

En conclusión, el análisis por país permite reconocer que las ventas registradas no se distribuyen de manera uniforme entre los mercados, sino que se encuentran concentradas principalmente en el Reino Unido.
